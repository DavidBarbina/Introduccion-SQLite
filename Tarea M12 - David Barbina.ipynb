```python
# Importamos la librerías correspondiente
import sqlite3
import pandas as pd
```

```python
# Creamos la db
conn = sqlite3.connect("Base.db")
```

```python
# Establecemos una estructura de control para manipular registros en una base de datos
cur = conn.cursor()
```

```python
# Creamos la tabla "Clientes"
cur.execute("""
    CREATE TABLE Clientes (
        CustomerID INTEGER PRIMARY KEY,
        CustomerName TEXT,
        ContactName TEXT,
        Address TEXT,
        City TEXT,
        PostalCode TEXT,
        Country TEXT,
        PurchaseUSD INT
        );
""")
```

```python
# Insertamos los primeros 5 registros

df = pd.read_csv("Cientifico de datos M40 Customers.csv") # Lectura
df_5 = df.head(5) # Selección de los primeros 5
df_5.to_sql("Clientes", conn, if_exists="append", index=False) # Insersión en la tabla "Clientes"
```

```python
# Generación del .sql con dump
with open("Operaciones.sql","w") as _file:
    for line in conn.iterdump():
        _file.write(f"{line}\n")
```

```python
# Mostramos los 5 registros
cur.execute("SELECT * FROM Clientes;") # Se seleccionan todos con el "SELECT *")
filas = cur.fetchall() # Se consideran todos

for fila in filas:
    print(fila)
```
